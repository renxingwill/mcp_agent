使用cherry studio作为客户端，配置小红书发布器

#### step1：下载安装cherry studio

#### step2：配置LLM

设置——模型服务——ModelScope魔搭——获取密钥，填入密钥即可完成魔搭社区的模型服务配置。

![](https://intranetproxy.alipay.com/skylark/lark/0/2025/png/26846/1744699656422-f9055673-9c4c-40ce-9e54-db1b40631dc1.png)

#### step3：配置小红书server和文生图server

设置——MCP服务器——编辑MCP配置

![](https://intranetproxy.alipay.com/skylark/lark/0/2025/png/26846/1744699656424-9e8a100b-bd91-45c2-85f3-32b055528b6f.png)

填入配置信息，点击确定




```python
{
  "mcpServers": {
    "modelscope-image-gen-mcp": {
        "command": "python",
        "args": [
            "-m",
            "modelscope_image_gen_mcp"
        ],
        "env": {
            "MODELSCOPE_TOKEN": "YOUR_MODELSCOPE_TOKEN" # KEY可以通过https://modelscope.cn/my/myaccesstoken获取
        }
    },
    "xiaohongshu": {
      "isActive": true,
      "command": "/opt/homebrew/Caskroom/miniforge/base/bin/python",  # 需要改成自己本地的python路径，可以打开终端，用which python命令获取
      "args": [
        "-m",
        "xhs_mcp_server"
      ],
      "name": "xiaohongshu"
    }
  }
}
```
